# Checkpoint 2: Verificação Formal de Tautologias e Modus Ponens
Demonstração computacional da regra de inferência [(P -> Q) & P] -> Q.


```python
import sympy as sp
from sympy.logic.boolalg import Implies, And

P, Q = sp.symbols('P Q')

# Modus Ponens: [(P -> Q) ^ P] -> Q
cond = Implies(P, Q)
premissas = And(cond, P)
modus_ponens = Implies(premissas, Q)

print("Proposição Modus Ponens:")
print(modus_ponens)

# Tabela verdade formal
print(f"\n{'P':<4} {'Q':<4} {'P->Q':<6} {'Prems':<6} {'Resultado'}")
print("-" * 35)
for p_val in [False, True]:
    for q_val in [False, True]:
        c = (not p_val) or q_val
        prem = c and p_val
        res = (not prem) or q_val
        print(f"{int(p_val):<4} {int(q_val):<4} {int(c):<6} {int(prem):<6} {int(res)}")

```

```python
import matplotlib.pyplot as plt

cenarios = ['P=0, Q=0', 'P=0, Q=1', 'P=1, Q=0', 'P=1, Q=1']
saidas = [1, 1, 1, 1]

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(cenarios, saidas, color='#34a853', width=0.4)
ax.set_ylabel('Valor Lógico Verdade', fontweight='bold')
ax.set_title('Validação de Tautologia Absoluta: Modus Ponens (S = 1)', fontweight='bold')
ax.set_ylim(0, 1.3)
ax.set_yticks([0, 1])
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
