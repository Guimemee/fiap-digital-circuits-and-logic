# Checkpoint 5: Somador Completo de 1 Bit (Full Adder)
Implementação das funções de soma S e transporte Cout da ULA.


```python
print(f"{'A':<3} {'B':<3} {'Cin':<4} | {'Soma S':<7} {'Cout':<5} | {'Decimal'}")
print("-" * 35)

for a in [0, 1]:
    for b in [0, 1]:
        for cin in [0, 1]:
            soma = a ^ b ^ cin
            cout = (a and b) or (cin and (a ^ b))
            val_dec = cout * 2 + soma
            print(f"{a:<3} {b:<3} {cin:<4} | {soma:<7} {cout:<5} | {val_dec}")

```

```python
import matplotlib.pyplot as plt

estados = ['000', '001', '010', '011', '100', '101', '110', '111']
decimais = [0, 1, 1, 2, 1, 2, 2, 3]

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(estados, decimais, color='#1f77b4', width=0.5)
ax.set_xlabel('Vetor de Entrada (A, B, Cin)', fontweight='bold')
ax.set_ylabel('Resultado Aritmético Decimal (0 a 3)', fontweight='bold')
ax.set_title('Resposta Aritmética do Somador Completo Binário (Full Adder)', fontweight='bold')
for b in bars:
    h = b.get_height()
    ax.text(b.get_x() + b.get_width()/2.0, h + 0.05, str(h), ha='center', fontweight='bold')
ax.set_ylim(0, 3.5)
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
