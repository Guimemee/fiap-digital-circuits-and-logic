# Checkpoint 3: Comparador de Magnitude Digital de 1 Bit
Lógica das saídas A>B, A=B e A<B para bancada de CIs TTL.


```python
print(f"{'A':<3} {'B':<3} | {'A > B':<7} {'A == B':<7} {'A < B':<7}")
print("-" * 30)

for a in [0, 1]:
    for b in [0, 1]:
        maior = int(a and not b)
        igual = int(a == b)
        menor = int(not a and b)
        print(f"{a:<3} {b:<3} | {maior:<7} {igual:<7} {menor:<7}")

```

```python
import matplotlib.pyplot as plt
import numpy as np

estados = ['A=0, B=0', 'A=0, B=1', 'A=1, B=0', 'A=1, B=1']
s_maior = [0, 0, 1, 0]
s_igual = [1, 0, 0, 1]
s_menor = [0, 1, 0, 0]

x = np.arange(len(estados))
w = 0.25

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(x - w, s_maior, w, label="A > B (A·B')", color='#ea4335')
ax.bar(x, s_igual, w, label="A == B (XNOR)", color='#4285f4')
ax.bar(x + w, s_menor, w, label="A < B (A'·B)", color='#fbbc05')

ax.set_ylabel('Nível Lógico de Saída', fontweight='bold')
ax.set_title('Respostas do Comparador Digital de Magnitude de 1 Bit', fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(estados)
ax.set_yticks([0, 1])
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
