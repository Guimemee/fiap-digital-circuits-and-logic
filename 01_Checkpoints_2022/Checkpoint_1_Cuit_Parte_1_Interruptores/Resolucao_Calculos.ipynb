# Checkpoint 1: Modelagem de Interruptores e Portas Lógicas Básicas
Simulação da tabela verdade da função de segurança S = A·B + C.


```python
import itertools

print(f"{'A':<3} {'B':<3} {'C':<3} | {'A·B':<5} | {'Saída S':<8}")
print("-" * 30)

estados = []
saidas = []
for a, b, c in itertools.product([0, 1], repeat=3):
    termo = a and b
    s = termo or c
    estados.append(f'{a}{b}{c}')
    saidas.append(s)
    print(f"{a:<3} {b:<3} {c:<3} | {termo:<5} | {s:<8}")

print(f'\nTotal de estados ativos: {sum(saidas)} de {len(saidas)}')

```

```python
import matplotlib.pyplot as plt

cores = ['#2ca02c' if s == 1 else '#d62728' for s in saidas]
fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(estados, saidas, color=cores, width=0.5)
ax.set_xlabel('Combinação das Entradas (ABC)', fontweight='bold')
ax.set_ylabel('Estado da Saída Lógica (0 ou 1)', fontweight='bold')
ax.set_title('Tabela Verdade da Função de Segurança S = (A·B) + C', fontweight='bold')
ax.set_yticks([0, 1])
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
