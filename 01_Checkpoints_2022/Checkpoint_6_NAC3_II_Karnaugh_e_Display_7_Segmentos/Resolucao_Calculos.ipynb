# Checkpoint 6: Decodificador BCD para Display de 7 Segmentos (Segmento 'a')
Minimização lógica por Mapa de Karnaugh de 4 variáveis com Don't Care no SymPy.


```python
import sympy as sp
from sympy.logic.boolalg import SOPform

A, B, C, D = sp.symbols('A B C D')
minterms = [0, 2, 3, 5, 6, 7, 8, 9]
dont_cares = [10, 11, 12, 13, 14, 15]

seg_a_min = SOPform([A, B, C, D], minterms, dont_cares)
print(f"Mintermos do Segmento 'a': {minterms}")
print(f"Condições Don't Care:     {dont_cares}")
print(f"Expressão Mínima SOP:     S_a = {seg_a_min}")

```

```python
import matplotlib.pyplot as plt

digitos = range(10)
seg_a_estado = [1, 0, 1, 1, 0, 1, 1, 1, 1, 1]

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(digitos, seg_a_estado, color=['#2ca02c' if s==1 else '#d62728' for s in seg_a_estado], width=0.5)
ax.set_xlabel('Dígito Decimal BCD (0 a 9)', fontweight='bold')
ax.set_ylabel('Estado do Segmento "a"', fontweight='bold')
ax.set_title('Comportamento Lógico do Segmento "a" no Display de 7 Segmentos', fontweight='bold')
ax.set_xticks(digitos)
ax.set_yticks([0, 1])
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
