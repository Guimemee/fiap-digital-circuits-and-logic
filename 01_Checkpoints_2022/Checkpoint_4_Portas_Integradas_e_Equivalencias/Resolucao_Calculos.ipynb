# Checkpoint 4: Universalidade de Portas Lógicas NAND
Comprovação de equivalência booleana da conversão De Morgan com SymPy.


```python
import sympy as sp
from sympy.logic.boolalg import Or, And, Nand

A, B, C, D = sp.symbols('A B C D')

# Circuito AND-OR
s_original = Or(And(A, B), And(C, D))

# Circuito All-NAND
nand_ab = Nand(A, B)
nand_cd = Nand(C, D)
s_nand = Nand(nand_ab, nand_cd)

print("Original (AND-OR):", s_original)
print("Convertido (NAND):", s_nand)

# Verificando equivalencia formal
print("As expressões são perfeitamente equivalentes?", sp.simplify_logic(s_original) == sp.simplify_logic(s_nand))

```

```python
import matplotlib.pyplot as plt

portas_antes = ['AND (2x)', 'OR (1x)']
qtd_antes = [2, 1]
portas_depois = ['NAND (3x)']
qtd_depois = [3]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))
ax1.bar(portas_antes, qtd_antes, color=['#1f77b4', '#ff7f0e'], width=0.4)
ax1.set_title('Antes: 2 CIs (7408 + 7432)')
ax1.set_ylabel('Quantidade de Portas')
ax1.set_ylim(0, 4)

ax2.bar(portas_depois, qtd_depois, color='#2ca02c', width=0.4)
ax2.set_title('Depois: 1 Único CI (7400 NAND)')
ax2.set_ylim(0, 4)

plt.suptitle('Otimização de Hardware via Universalidade de Portas NAND', fontweight='bold')
plt.tight_layout()
plt.show()

```
