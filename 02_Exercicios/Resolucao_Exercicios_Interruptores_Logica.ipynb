### Convenções Normativas de Circuitos de Interruptores:
1. **Associação em Série:** Operação Lógica Conjunção (`AND` / $\cdot$ / $\wedge$)
2. **Associação em Paralelo:** Operação Lógica Disjunção (`OR` / $+$ / $\vee$)
3. **Interruptor Normal Aberto (NA):** Variável afirmativa ($p, q, r$)
4. **Interruptor Normal Fechado (NF):** Variável negada ($\sim p, \sim q, \sim r$)

---


```python
import pandas as pd
import numpy as np
import sympy as sp
from sympy.logic.boolalg import simplify_logic

# Função utilitária para gerar tabela da verdade estilizada em Pandas
def gerar_tabela_verdade_2var(func_lambda, nome_func):
    dados = []
    for p in [0, 1]:
        for q in [0, 1]:
            saida = int(func_lambda(bool(p), bool(q)))
            dados.append({'p': p, 'q': q, '~p': 1-p, '~q': 1-q, nome_func: saida})
    df = pd.DataFrame(dados)
    return df

def gerar_tabela_verdade_3var(func_lambda, nome_func):
    dados = []
    for p in [0, 1]:
        for q in [0, 1]:
            for r in [0, 1]:
                saida = int(func_lambda(bool(p), bool(q), bool(r)))
                dados.append({'p': p, 'q': q, 'r': r, '~p': 1-p, '~q': 1-q, '~r': 1-r, nome_func: saida})
    df = pd.DataFrame(dados)
    return df
```

### Exercício 1 (Item a):
- **Esquema de Interruptores:** Ramo superior com $p$ em série com $\sim q$; ramo inferior com $\sim p$ em série com $q$.
- **WFF (Well-Formed Formula):** $S_a = (p \wedge \sim q) \vee (\sim p \wedge q) = p \cdot \overline{q} + \overline{p} \cdot q$
- **Simplificação Distributiva:** Corresponde exatamente à função **XOR (Ou-Exclusivo)**: $S_a = p \oplus q$.

```python
df_a = gerar_tabela_verdade_2var(lambda p, q: (p and not q) or (not p and q), 'Sa = (p∧~q) ∨ (~p∧q)')
print('Tabela Verdade - Item (a):')
display(df_a)
```

### Exercício 1 (Item b):
- **Esquema de Interruptores:** $(p \vee q) \wedge (\sim p \vee \sim q)$
- **WFF:** $S_b = (p + q) \cdot (\overline{p} + \overline{q})$
- **Aplicação da Distributiva:**
  $$S_b = p \cdot \overline{p} + p \cdot \overline{q} + q \cdot \overline{p} + q \cdot \overline{q} = 0 + p \cdot \overline{q} + \overline{p} \cdot q + 0 = p \cdot \overline{q} + \overline{p} \cdot q = p \oplus q$$
- **Conclusão:** É equivalente ao item (a).

```python
df_b = gerar_tabela_verdade_2var(lambda p, q: (p or q) and (not p or not q), 'Sb = (p∨q) ∧ (~p∨~q)')
print('Tabela Verdade - Item (b):')
display(df_b)
```

### Exercício 1 (Item c ao Desafio):
- **Item c:** $S_c = p \cdot (q + \overline{q}) = p \cdot 1 = p$
- **Item d:** $S_d = (p + q) \cdot (p + \overline{q}) = p + (q \cdot \overline{q}) = p + 0 = p$
- **Desafio (3 Variáveis):** $S_{\text{desafio}} = (a \cdot b) + (\overline{a} \cdot c \cdot d) + (e \cdot \overline{e}) + (c \cdot a \cdot \overline{e}) = a \cdot b + \overline{a} \cdot c \cdot d + a \cdot c \cdot \overline{e}$.

```python
# Verificação Simbólica via SymPy de Todas as Simplificações
p, q, r, a, b, c, d, e = sp.symbols('p q r a b c d e')

expr_a = (p & ~q) | (~p & q)
expr_b = (p | q) & (~p | ~q)
expr_d = (p | q) & (p | ~q)

print('Simplificação Item (a):', simplify_logic(expr_a))
print('Simplificação Item (b):', simplify_logic(expr_b))
print('Simplificação Item (d):', simplify_logic(expr_d))
```
